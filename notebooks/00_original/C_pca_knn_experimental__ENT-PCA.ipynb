```python
# Reference Paper - https://arxiv.org/pdf/2203.09610.pdf
'''
def FS_ASD(k):

    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)

    if k == 0 :
        psi[0] = np.random.random()
    elif k == 1 :
        psi[6] = np.random.random()
    elif k == 2 :
        psi[5] = np.random.random()
    else :
        return None

    #psi = np.array(map(lambda x: x/np.linalg.norm(psi), psi))
    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))
    return rho.data.toarray()
'''

def FS_ASD():
    k = random.choice([0,1,2])
    #k= 0
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)

    if k == 0 :
        psi[0] = np.random.random()
    elif k == 1 :
        psi[6] = np.random.random()
    elif k == 2 :
        psi[5] = np.random.random()
    else :
        return None

    #psi = np.array(map(lambda x: x/np.linalg.norm(psi), psi))
    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))
    return rho.data.toarray()

def BS1_ASD():
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi[5] = (-1)*np.random.default_rng().uniform(-1,0)
    psi[6] = (-1)*np.random.default_rng().uniform(-1,0)
    psi[7] = (-1)*np.random.default_rng().uniform(-1,0)

    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))
    return rho.data.toarray()

def BS2_ASD():
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8

    psi[0] = (-1)*np.random.default_rng().uniform(-1,0)               # 000
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)# 100
    psi[5] = (-1)*np.random.default_rng().uniform(-1,0)               # 101
    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))

    return rho.data.toarray()

def BS3_ASD():
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8

    psi[0] = (-1)*np.random.default_rng().uniform(-1,0)               # 000
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)# 100
    psi[6] = (-1)*np.random.default_rng().uniform(-1,0)               # 110
    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))

    return rho.data.toarray()

def W_ASD():
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8

    psi[0] = (-1)*np.random.default_rng().uniform(-1,0)               # 000
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)# 100
    psi[5] = (-1)*np.random.default_rng().uniform(-1,0)               # 101
    psi[6] = (-1)*np.random.default_rng().uniform(-1,0)               # 110
    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))

    return rho.data.toarray()

def GHZ_ASD():
    phi = (2*np.pi)*(-1)*np.random.default_rng().uniform(-1,0)
    psi = [0+0*1j]*8

    psi[0] = (-1)*np.random.default_rng().uniform(-1,0)               # 000
    psi[4] = np.exp(1j*phi)*(-1)*np.random.default_rng().uniform(-1,0)# 100
    psi[5] = (-1)*np.random.default_rng().uniform(-1,0)               # 101
    psi[6] = (-1)*np.random.default_rng().uniform(-1,0)               # 110
    psi[7] = (-1)*np.random.default_rng().uniform(-1,0)               # 111

    psi = np.array(psi/np.linalg.norm(psi))
    rho = ket2dm(Qobj(psi, dims = [[2,2,2],[1,1,1]]))

    return rho.data.toarray()

def three_tangle_ASD(psi):
    t000 = psi[0]
    t111 = psi[7]
    tau_abc = 4*(t000**2)*(t111**2)
    return tau_abc

def array_to_label_list(Matrix,n,k):
    '''
    Matrix refers to the density matrix which is supposed to be input in this function,
    of whose the list is to be constructed

    n = Number of qubits in the system

    k = index reference to the three qubit entanglement class
        k = 0 => FS => Full Separable => A-B-C => [1,0,0,0,0,0]
        k = 1 => bs2 => Bi-Separable-1 => A-BC => [0,1,0,0,0,0]
        k = 2 => BS2 => Bi-Separable-2 => AB-C => [0,0,1,0,0,0]
        k = 3 => BS3 => Bi-Separable-3 => AC-B => [0,0,0,1,0,0]
        k = 4 => W => Werner State => ABC =>      [0,0,0,0,1,0]
        k = 5 => w => Greenberg-Horne-Zeilinger State => ABC => [0,0,0,0,0,1]

    From 0 to 31 elements of the returned list real components of the density matrix(Re(rho_00),Re(rho_01),...) i.e, from 0 to N-1
    from 32 to 63 : Imaginary components of the density matrix i.e, (Im(rho_00),Im(rho_01),...)
    from 64 to 69 : One hot encoded labels corresponding to the class of the 3 qubit entangled states
    '''
    N = 2**(2*n)
    Matrix_1 = Matrix.reshape((1,N))
    #data = np.zeros((2*N))
    data = [0]*(2*N)
    FS  = [1,0,0,0,0,0,0]
    bs2 = [0,1,0,0,0,0,1]
    BS2 = [0,0,1,0,0,0,2]
    BS3 = [0,0,0,1,0,0,3]
    W   = [0,0,0,0,1,0,4]
    GHZ = [0,0,0,0,0,1,5]

    for i in range(0,N):
        data[i] = np.real(Matrix_1[0][i])
    for j in range(N,2*N):
        data[j] = np.imag(Matrix_1[0][j-N])

    #data  = data.tolist()
    if k == 0:
        data.extend(FS)
    elif k == 1:
        data.extend(bs2)
    elif k == 2:
        data.extend(BS2)
    elif k == 3:
        data.extend(BS3)
    elif k == 4:
        data.extend(W)
    elif k == 5:
        data.extend(GHZ)
    else :
        return "Error : Corresponding state label doesn't exist, select from 0,1,2,3,4 or 5"
    return data

def labels(n):

    N=2**(2*n)
    fields = [None]*((2*N))
    list_classes_1 = ['FS','BS1','BS2','BS3','W','GHZ','CODE']

    k = 0
    for i in range(2**n):
        for j in range(2**n):
            fields[k] = 'Rho_'+str(i)+'_'+str(j)+'_Real'
            k=k+1

    k = N
    for i in range(2**n):
        for j in range(2**n):
            fields[k] = 'Rho_'+str(i)+'_'+str(j)+'_Imag'
            k=k+1
    fields.extend(list_classes_1)
    return fields

def class_return_3q(k):
    if k == 0:
        return FS_ASD()
    elif k == 1:
        return BS1_ASD()
    elif k == 2:
        return BS2_ASD()
    elif k == 3:
        return BS3_ASD()
    elif k == 4:
        return W_ASD()
    elif k == 5:
        return GHZ_ASD()
    else :
        return "Class index out of [0,1,2,3,4,5]"

def data_generator_3q_class(num, k):
    data_list = []

    '''
    K = The type of entanglement class one would want to generate with this
            function
    k = 0 => FS => Full Separable => A-B-C => [1,0,0,0,0,0]
    k = 1 => bs2 => Bi-Separable-1 => A-BC => [0,1,0,0,0,0]
    k = 2 => BS2 => Bi-Separable-2 => AB-C => [0,0,1,0,0,0]
    k = 3 => BS3 => Bi-Separable-3 => AC-B => [0,0,0,1,0,0]
    k = 4 => W => W State => ABC =>      [0,0,0,0,1,0]
    k = 5 => w => Greenberg-Horne-Zeilinger State => ABC => [0,0,0,0,0,1]
    Num = No. of quantum states of the class required
    '''

    for i in range(num):
        Matrix = class_return_3q(k)
        data = array_to_label_list(Matrix=Matrix,n=3,k=k)
        data_list.append(data)
    return data_list

def dim_reduce(dataframe):

    #index_list = [0,1,2,3,4,5,6,7]
    col_name_list = []
    col_list = []

    for i in range(0,8):
        for j in range(i,8):
            col_name = 'Rho_'+str(i)+'_'+str(j)+'_Real'
            col_name_list.append(col_name)
            col_list.append(dataframe[col_name])

    for i in range(0,8):
        for j in range(i,8):
            col_name = 'Rho_'+str(i)+'_'+str(j)+'_Imag'
            col_name_list.append(col_name)
            col_list.append(dataframe[col_name])

    col_list.append(dataframe[['FS','BS1','BS2','BS3','W','GHZ','CODE']])
    df = pd.concat(col_list, axis = 1)

    return df

def dim_reduce_1(dataframe):

    #index_list = [0,1,2,3,4,5,6,7]
    col_name_list = []
    col_list = []

    for i in range(0,8):
        for j in range(i,8):
            col_name = 'Rho_'+str(i)+'_'+str(j)+'_Real'
            col_name_list.append(col_name)
            col_list.append(dataframe[col_name])

    for i in range(0,8):
        for j in range(i,8):
            col_name = 'Rho_'+str(i)+'_'+str(j)+'_Imag'
            col_name_list.append(col_name)
            col_list.append(dataframe[col_name])

    diag_imag_list = ['Rho_'+str(i)+'_'+str(i)+'_Imag' for i in range(8)]

    col_list.append(dataframe[['FS','BS1','BS2','BS3','W','GHZ','CODE']])
    df = pd.concat(col_list, axis = 1)
    df = df.drop(diag_imag_list, axis = 1 )

    return df
```
